```python
import pandas as pd
import pm4py
from pathlib import Path

INPUT = Path("./../../MIMICEL_data/mimicel_train.csv")
OUT = Path("./results/mimicel_train.xes")

OUT.parent.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(INPUT)

df = df.rename(columns={
    "stay_id": "case:concept:name",
    "activity": "concept:name",
    "timestamps": "time:timestamp",
    "timestamp": "time:timestamp"
})

df = df[["case:concept:name", "concept:name", "time:timestamp"]].copy()

df["case:concept:name"] = df["case:concept:name"].astype(str)
df["concept:name"] = df["concept:name"].astype(str)
df["time:timestamp"] = pd.to_datetime(df["time:timestamp"], errors="coerce")

df = df.dropna(subset=["time:timestamp"])
df = df.sort_values(["case:concept:name", "time:timestamp"]).reset_index(drop=True)

event_log = pm4py.convert_to_event_log(
    df,
    case_id_key="case:concept:name"
)

pm4py.write_xes(event_log, str(OUT))

print("saved:", OUT)
print("cases:", df["case:concept:name"].nunique())
print("events:", len(df))
print(df.dtypes)
print(df.head())
```
