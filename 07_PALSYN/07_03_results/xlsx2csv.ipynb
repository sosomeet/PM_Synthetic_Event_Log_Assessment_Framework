```python
from pathlib import Path
import pandas as pd
import re

INPUT_DIR = Path("./")
OUTPUT_DIR = Path("./result")
OUTPUT_DIR.mkdir(exist_ok=True)

xlsx_files = sorted(INPUT_DIR.glob("palsyn*.xlsx"))

for xlsx_file in xlsx_files:
    match = re.search(r"(\d+)", xlsx_file.stem)
    if not match:
        print(f"[SKIP] seed 번호 없음: {xlsx_file.name}")
        continue

    seed = match.group(1)

    df = pd.read_excel(xlsx_file)
    output_path = OUTPUT_DIR / f"palsyn_synthetic_seed{seed}.csv"

    df.to_csv(output_path, index=False, encoding="utf-8-sig")
    print(f"[OK] {xlsx_file.name} -> {output_path}")

print("완료")
```
